```python
from game.public_goods_game import PublicGoodsGame
import pandas as pd

results = []
strategy_combinations = [
    {"coop": 10, "defect": 10},
    {"coop": 10, "defect": 10, "random": 10},
    {"adaptive": 20, "defect": 10},
    {"adaptive": 15, "coop": 5, "defect": 10},
]

for strategies in strategy_combinations:
    game = PublicGoodsGame.run_simulation(
        turns=50,
        endowment=20,
        factor=2.0,
        strategy=strategies,
        width=8,
        height=8,
        num_neighborhoods=4,
        local_game=True,
        councils=False,
        show_stats=False,
        show_map=False
    )
    final_coop = game.history[-1]['average_cooperation']
    final_payoff = game.history[-1]['average_payoff']
    results.append({
        'strategies': strategies,
        'final_cooperation': final_coop,
        'final_payoff': final_payoff
    })

df = pd.DataFrame(results)
print(df.sort_values('final_cooperation', ascending=False))
```
